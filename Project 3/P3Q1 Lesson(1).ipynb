```python
import pandas as pd # Importing pandas library for data pre-processing 
myDF = pd.read_csv("/anvil/projects/tdm/data/flights/subset/airports.csv") # Reading csv file into myDF
myDF.head() # Displays top 5 results of the csv, by chronological order
# We see first row (row 0) has Thigpen as the airport, Bay Springs as the city, Mexico State as the state, USA as the country,
# 31.95... for latitude, -89.23... for longitutude
```

```python
myDF['state'].head()
# If we sort by state, we just get the state name itself
```

```python
myDF['state'].value_counts().head() # One interesting observation can be made here that is that python's value_counts function
# has a built in sorting function. This is further proven in the documentation of the pandas library.

# Just a short comment on the results itself we can see that Alaska has the most number of airports of 263 total airports, followed
# by Texas with 209, California with 205, Oakland with 102, and Ohio with 100.
```

```python
akDF = myDF[(myDF['state'] == 'AK')] # Defines akDF as a data frame with just alaska as the state
# Note: if you just did "(myDF['state'] == 'AK')" it would just be a table of true and false which is not helpful
#, so specifying the condition on the dataset is quite important.

akDF.head() # shows the top 5 alaska airports, sorted by chronological order
```

```python
import matplotlib.pyplot as plt # Importing matplotlib library for plotting bar plot
myStateCounts = myDF['state'].value_counts().head() # Defining a new dataframe for the top 5 results of number of airports
# sorted by the state they are located in. Reminder of what is displayed is above
plt.bar(myStateCounts.index,myStateCounts) # The index of this table is quite interesting. Since it is sorted by state, then 
# state actually BECOMES the index of the x-axis 
```
