```python
correctNonStateDF <- subset(myZillowDF, !(RegionName %in% gsub(" ", "", state.name))) # We use the gsub function that replaces all 
# instances of space bar in a specific String (in this case, the state.name character vector) and instead replace with no spaces
# so that we are actually comparing the rights things and not rejecting actual state results. One thing to note is that an alternate
# solution I thought of was to create a loop for all results in RegionName to add spaces right before a character that is capitalized,
# but I recognize this takes more computation and time although it could help in the long run where clear display may be needed.
# There may also be another issue where, for example, in "DistrictofColumbia" has of in the middle of the string which isn't capitalized
# but it is still a separate word. A possible solve to this would be to match each word with a dictionary or use a natural language
# processing library to see the RegionName we get after adding spaces makes sense or not.

head(correctNonStateDF) # Shows top 6 results of the corrected non-state dataframe, and we can see that now we actually get correct
# results of having results with regions like district of columbia rather than also including results with regions like new york.

dim(correctNonStateDF) # Displaying dimensions of the "correctNonStateDF" dataframe which correctly displays that there are 328 rows/
# results and that there are 82 columns in the subset.
```

```python
import pandas as pd # Importing the pandas library for csv reading, dataset pre-processing and basic analysis

myTaxiDF = pd.read_csv("/anvil/projects/tdm/data/taxi/green/green_tripdata_2019-06.csv") # Reads the green_tripdata in 2019 june csv
# into a dataframe called "myTaxiDF" through panda's read_csv function

myTaxiDF.head()# Displays the top 5 results of the myTaxiDF, sorted chronologically, to give us a visualization and basic understanding
```

```python
countOfLocationDF = myTaxiDF['DOLocationID'].value_counts() # Creating a new dataframe called "countOfLocationDF" that counts the number
# of occurences for each DOLocationID.

moreThanTenkIndexDF = countOfLocationDF[countOfLocationDF > 10000] # We want to create a new dataframe that isolates "countOfLocationDF"
# instances that have a count of how often a specific DOLocationID occured that is larger than 10000.

print(moreThanTenkIndexDF) # We can print the value_counts series to see all DOLocationID that have an occurence larger than 10000. From
# the result, we can see that DOLocationID of 74 has a occurence of 19187 times, 42 has an occurence of 17378 times, 41 has an occurence
# of 15547 times, 75 has an occurnece of 13270 times, 7 has an occurence of 13035 times, 129 has an occurence of 12966 times, and 
# 166 has an occruence of 10389 times. This matches what we expected of a total of 7 values of DOLocationID that occur 10000+ times.
```

```python
onlyOnePassengerDF = myTaxiDF["passenger_count"] # Creating the "onlyOnePassengerDF" dataframe that sorts by how often a specific 
# passenger_count shows up (for example, how often 1 passenger count taxis occured)

print(onlyOnePassengerDF.value_counts().sort_index()) # We want to return the count of how many rows have passenger count of a speciic
# number. More specifically, by having it sorted by index, we have a cleaner analysis. From this, we can notice that there are 
# 403013 instances/rows/results that have only 1 passenger in the taxi while there was only 36897 instances/rows/results that have 
# 2 passengers. This does make quite a lot of sense as most people do solo taxi while groups of 5 also show up quite a lot, potentially
# because it is the max capacity of most taxis and most people need taxis after parties or gatherings.
```

```python
countOfMoneyDF = myTaxiDF[myTaxiDF['tip_amount'] >= 21] # Creating "countOfMoneyDF" dataframe that only accepts rows/results/instances
# that have a tip amount that is larger or equal to 21 dollars. 

countOfMoneyDF.shape # By getting the shape of countOfMoney, we get 216 rows and 20 columns. The result we have of interest is the 
# fact that it has 216 rows/instances/results that have a tip larger or equal to 21 dollars.
```

```python
myAccidentDF = pd.read_csv("/anvil/projects/tdm/data/fars/2021/accident.csv", encoding="latin1") # Reading the accidents.csv file 
# into the "myAccidentDF" dataframe using panda's read_csv function. Encoding = "latin1" is used here because there is some irregular
# encoding in the dataset, in which if UTF-8 was used, it would lead to some reading errors as stated by the error code: 
# 'utf-8' codec can't decode byte 0xd1 in position 189939: invalid continuation byte

myAccidentDF.head() # Displaying top 5 results, sorted chronologically to identify the column of states that should be targeted
```

```python
myAccidentStateDF = myAccidentDF['STATENAME'].value_counts() # This is just for visualization purposes, but I am creating the 
# "myAccidentStateDF" dataframe that counts how often each STATENAME occured in the dataset by using value_counts function.

print(myAccidentStateDF) # Printing it out for visualization purposes.
```

```python
states = [
    "Alabama", "Alaska", "Arizona", "Arkansas", "California",
    "Colorado", "Connecticut", "Delaware", "Florida", "Georgia",
    "Hawaii", "Idaho", "Illinois", "Indiana", "Iowa",
    "Kansas", "Kentucky", "Louisiana", "Maine", "Maryland",
    "Massachusetts", "Michigan", "Minnesota", "Mississippi", "Missouri",
    "Montana", "Nebraska", "Nevada", "New Hampshire", "New Jersey",
    "New Mexico", "New York", "North Carolina", "North Dakota", "Ohio",
    "Oklahoma", "Oregon", "Pennsylvania", "Rhode Island", "South Carolina",
    "South Dakota", "Tennessee", "Texas", "Utah", "Vermont",
    "Virginia", "Washington", "West Virginia", "Wisconsin", "Wyoming"
] # Initializing all the state names in python because there is no built in list for us to access. 

mask = myAccidentDF["STATENAME"].isin(states) # Was originally going to loop through all indices in the array and the dataframe, 
# but it would take too long and isn't as efficient. There is a very convenient function in pandas called "isin" that checks whether
# a dataframe column result is in a specific array/result. 

print(mask) # By printing out the mask, we can see that it is a list of trues and falses with corresponding rows in the myAccidentDF.
# This is not extremely useful in terms of visualization, but with some extra processing, we can get some really meaningful data.
```

```python
nonMatchesDF = myAccidentDF[~mask] # We are initializing "nonMatchesDF" dataframe which takes all the results that have states that 
# are NOT inside the states array. We achieve this by using the "tilde" sign in pandas that helps to invert results in the mask
# and instead, our ~mask would get us something like false, false, false, false ... if printed out. 

print(nonMatchesDF["STATENAME"].value_counts()) # The final step here is to print out the count of the instances in which a state
# that is not inside the states array, from this we can observe that there are 40 instances/rows/results that have District of Columbia
# as their "STATENAME".
```
