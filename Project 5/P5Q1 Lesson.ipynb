```r
library(data.table)
myDF <- fread("/anvil/projects/tdm/data/beer/breweries.csv")
```

```r
head(myDF)
```

```r
head(myDF$state)
```

```r
head(myDF[myDF$state %in% state.abb, ])

# Displays top 6 valid results that have a valid US state that is inside the state.abb character vector, suggesting that non-us and 
# non-valid states are excluded from the result
```

```r
head(myDF[myDF$country == "US", ])
```

```r
dim(myDF[myDF$state %in% state.abb, ])
dim(myDF[myDF$country == "US", ])
# As we can see there are some results that have US as the country, but don't have a state inside the state.abb character vector
```

```r
head(myDF[(myDF$country == "US") & (!(myDF$state %in% state.abb)), ])
# This returns top 6 results that have US country but the opposite of what we did before, which is that that the state in a row
# does not exist within the state.abb vector as seen from empty states or DC
```

```r
dim(myDF[(myDF$country == "US") & (!(myDF$state %in% state.abb)), ])
# Finds how many rows that are US but not valid state in state.abb
```

```r
head(subset(myDF, (country == "US") & (!(state %in% state.abb))))
```

```r
dim(subset(myDF, (country == "US") & (!(state %in% state.abb))))
```
