```python
myBikeDF <- fread("/anvil/projects/tdm/data/bay_area_bike_share/baywheels/202507-baywheels-tripdata.csv") # Reading the baywheels biking
# dataset into the "myBikeDF" dataframe by using the fread function which is more efficient than other forms of default reading in R

head(myBikeDF) # Shows the top 6 results of the myBikeDF dataframe, sorted chronologically by index of the dataset. No significant 
# result can be commented on at the moment, but it is important to recognize that we now have more clear visualization of what the 
# columns of the dataset look like and what the data type and formatting of the different columns look like. The two columns of note
# will probably be started_at and ended_at column which contain date data. One thing to note here is that the format of this column
# is slightly different as it also contains the date, but it also contains info about the hour, minute, and second it started and ended
```

```python
# My first approach did not have both start and end july dataframes, but this change was made for the detail I noticed.
# This is the implementation using start time column in the myBikeDF dataframe:
# --------------------------------------------------------------------------------------------------------------------------------------
myStartJulyDF <- subset(myBikeDF, year(myBikeDF$started_at) == 2025 & month(myBikeDF$started_at) == 7) # Defining a subset of "myBikeDF" 
# as "myStartJulyDF" for instances that started in July, 2025. When first aproaching this question and this dataset in general, I 
# thought it would be quite simple: I would just need to find the interval between two dates but I've noticed some nuance here. 

dim(myStartJulyDF)# Checking the dimensions of the "myStartJulyDF" dataframe. We get a result of 388142 rows/results and 13 columns.

myStartIntervalDF <- myStartJulyDF$ended_at - myStartJulyDF$started_at # We subtract the end and start time of the "myStartJuly" 
# dataframe and read it into the "myStartIntervalDF" dataframe since the date and time are already formatted really well.

meanStartInterval <- mean(myStartIntervalDF) # Defining a new variable called "meanStartInterval" that finds the mean of 
# "myStartIntervalDF" dataframe

print(meanStartInterval) # Outputting the results from the meanStartInterval process. We get "Time difference of 789.5851 secs".


# Below is the implementation using end time, rather than using start time column in the myBikeDF dataframe:
# --------------------------------------------------------------------------------------------------------------------------------------
myEndJulyDF <- subset(myBikeDF, year(myBikeDF$ended_at) == 2025 & month(myBikeDF$ended_at) == 7) # Defining a subset of myBikeDF as 
# "myEndJulyDF" for instances that ended in July, 2025.

dim(myEndJulyDF)# Checking the dimensions of the "myEndJulyDF" dataframe. We get a result of 388168 rows/results and 13 columns.

myEndIntervalDF <- myEndJulyDF$ended_at - myEndJulyDF$started_at # We subtract the end and start time of the "myEndJuly" dataframe 
# and read it into the "myEndIntervalDF" dataframe since the date and time are already formatted really well.

meanEndInterval <- mean(myEndIntervalDF) # Defining a new variable called "meanEndInterval" that finds the mean of "myEndIntervalDF" 
# dataframe

print(meanEndInterval) # Outputting the results from the meanEndInterval process. We get "Time difference of 791.8954 secs".


# As mentioned above, I did not know that a simple task like finding interval of time would require such nuanced decisions to be made
# prior to that actual process. The main thing here is that, because the question of the task only asks us to "Find the average length 
# of bicycle trips in July 2025," we get no extra information on whether we should use the start or end time or a mix of both when 
# trying to create a subset off of "myBikeDF" dataframe. This is obviously an issue as it leads to discrepancy in the answers to a 
# marginal extent. This creates a difference of 26 rows which is probably caused by rides that started in late June but ended in July.
# They are in the files, but they are excluded when filtering by started_at. Rides that started in July and ended in August are not in 
# this file at  all, so the start-based subset are rides that both started and ended in July.

# As explained later down below, because the assignemnet says the dataset contains 388168 rides, it matches the end in july set, the
# average length of the bike rides is 791.8954 secs. The start-based value of 789.5851 secs is shown only to illustrate the ambiguity.
```

```python
# Below is the implement using time difference of the entire dataset
# --------------------------------------------------------------------------------------------------------------------------------------
averageTimeDif <-mean(myBikeDF$ended_at - myBikeDF$started_at) # Defining "averageTimeDif" variable as the average time of time interval
# of all 388168 instances of bike rides in July 2025

myLargerMeanDF <- subset(myBikeDF, (myBikeDF$ended_at - myBikeDF$started_at)> averageTimeDif) # Creating a subset that catches
# all the instances in which a bike had an interval higher than the "averageTimeDif." 

dim(myLargerMeanDF) # Gets the dimensions of myLargerMeanDF dataframe. It gives 116823 rows and 13 columns.


# Below is the implementation using meanStartInterval as the average duration of bike ride
# --------------------------------------------------------------------------------------------------------------------------------------
myStartLargerMeanDF <- subset(myBikeDF, (myBikeDF$ended_at - myBikeDF$started_at)> meanStartInterval) # Creating a subset that catches
# all the instances in which a bike had an interval higher than the "meanStartInterval." 

dim(myStartLargerMeanDF) # Finding the dimensions of "myStartLargerMeanDF" dataframe. We get 117373 rows and 13 columns. 


# --------------------------------------------------------------------------------------------------------------------------------------

myEndLargerMeanDF <- subset(myBikeDF, (myBikeDF$ended_at - myBikeDF$started_at)> meanEndInterval) # Creating a subset that catches
# all the instances in which a bike had an interval higher than the "meanEndInterval." 

dim(myEndLargerMeanDF) # Finding the dimensions of "myEndLargerMeanDF" dataframe. We get 116823 rows and 13 columns. 

# The calculation here actually doesn't matter as much as the pre-processing we did to obtain the average of bike rides in July 2025,
# nevertheless, we get a either 117373 or 116823 people/instances that are longer than the average length. This is a difference of 
# 550 bike rides. This can be used to the company's advantage. For example, they could have a promotion that if you use a bike more than
# the average length, you get a discount, so they could try to encourage longer rentals by setting the average length calculated with
# ended_at in July 2025. One extra observation is that dim(myLargerMeanDF) matches dim(myEndLargerMeanDF) exactly, suggesting that 
# the full dataset of 388168 rides is already the set of rides that ended in july 2025, so averageTimeDif and meanEndInterval are the 
# same value.

# To answer the question more directly, the number of bike rides longer than the average length of rides is 116823 rows/rides. This is 
# once again due to the fact that the dataset naturally excludes rides that started in July and ended in August.
```

```python
import pandas as pd # Importing pandas library as pd (abbreviation)
import matplotlib.pyplot as plt # Importing matplotlib's pyplot function as plt (abbreviation)

myIceCreamPdDF = pd.read_csv("/anvil/projects/tdm/data/icecream/combined/reviews.csv") # Reading ice cream reviews data 

myIceCreamPdDF["date"].value_counts() # Displaying the counts of how often each date in which a review is written shows up, sorted
# in descending order (in terms of the value, not the index). 

# The answer here also matches what we got from R. The date in which the most reviews was written is "2020-09-21" or 2020, 
# September 21st with 362 instances of reviews on that day. Not much can be commented on at the moment because it isn't visualized
# in a meaningful way at the moment.
```

```python
myIceCreamPdDF["date"].value_counts().sort_index().plot() # Getting the date column in myIceCreamPdDF dataframe and getting the count
# of how often each date showed up using the value_counts function, similar to what we did in the previous step. 

# This matches what we saw above as 2020-09-21 was the most popular with 362 reviews then 2020-07-26 with 213 reviews which is followed
# closely by 2018-06-04 with 209 reviews. There is no clear trend here besides that there are couple spikes on specific dates and 
# spikes happening a bit more frequently around the time the data collection stopped, probably because more and more users joined
# the app. One thing that we can notice is that python's x-axis is extremely squished and quite unclear which is likely due to the 
# fact that is not scaled properly and the fact that there is a set container for matplotlib results unless otherwise stated. This
# suggests that further processing may be needed to extract a clear timestamp of when major trends and data points take place.
```

```python
myJune2018DF = myIceCreamPdDF[
(pd.to_datetime(myIceCreamPdDF["date"]).dt.year == 2018)
& (pd.to_datetime(myIceCreamPdDF["date"]).dt.month == 6)] # Creating another dataframe that has only results from 2018 June. We do this
# by first calling the to_datetime function in pandas that formats the date column into something usable, then we call the .dt.month
# function to extract and compare the month and the .dt.year function to extract and compare the year

myJune2018DF["date"].value_counts().sort_index().plot() # Getting the date column in myJune2018DF dataframe and getting the count
# of how often each date showed up using the value_counts function, similar to what we did in the previous step. However, since plots
# have the x-axis as the date, it would be much less meaningful if the data was sorted by value_counts since we can not observe a 
# clear trend over time. Therefore, we need to use the sort_index function before we plot so we get a plot sorted by time.

# The result is similar to what was observed in question 1, except for that pandas has the y-labels contained in a box so that all
# values are scaled to the dimension of the height of the box. From the graph, we can see the biggest spike of 209 on 2018-06-04,
# which is followed by 2018-06-11 with 110 reviews. Overall, the # graph is skewed left with spikes to the left, but no significant 
# trend can be inferred besides the fact that it could be possible  that most people post reviews at the start of the month either due 
# to promotional activities or just by pure accident. One thing to be noted is that pandas return the plot with a connected line
# unlike R which kinda does it like a bar chart, this could be a setting that I can explore in the future and evaluate.
```
