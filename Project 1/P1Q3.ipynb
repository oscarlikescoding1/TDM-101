```r
library(data.table) # Importing the data.table library through calling the library function
beerDF <- fread("/anvil/projects/tdm/data/beer/beers.csv") # A different and "more efficient" method of loading dataset
```

```r
tail(sort((table(beerDF$state))),n=6) # A sorted view of all beers, filtered by state of origin, with top 6 results shown
# In this case, it shows the top 6 results sorted by number of beer in specific state e.g. new york has 14,572 beers
```
