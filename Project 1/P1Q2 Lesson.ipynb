```python
import pandas as pd # Importing the pandas library to be able to read csv files and access other neat features
myDF = pd.read_csv("/anvil/projects/tdm/data/olympics/athlete_events.csv") # This calls the pandas "read_csv" function and creates
# a new dataframe that reads the dataset into the dataframe "myDF"
myDF.head() # Same thing in R, it shows the top results for the dataset (for number of rows, argument is just the number e.g. 12)
# Default n value is 5
```

```python
myWinterDF = myDF[(myDF['Season'] == 'Winter')] # This is the dataframe where we filter based on season when it's winter
# Just a quick note, the notation is [()] and parameteres inside (i feel like most python thingys are like this)
myWinterDF.shape # Obviously, we can see the size of the dataset shrinks to 48,564 athletes/rows
```

```python
myWinterPolandDF = myDF[(myDF['Season'] == 'Winter') & (myDF['NOC'] == 'POL')] # This is the dataframe in which
# we filter based on both the season when it's winter AND when NOC is "POL (or poland)", yet again it shrinks
myWinterPolandDF.shape # Displays the dimensions of the "myWinterPolandDF" dataset (1342 rows, 15 columns)
```

```python
myJamaicaBob1988DF = myDF[(myDF['NOC'] == 'JAM')& (myDF['Sport'] == 'Bobsleigh') & (myDF['Games'] == '1988 Winter')]
# Dataframe of athletes with filters of Jamaican, bobsleigh, and 1988 winter olympics
myJamaicaBob1988DF.shape # Shows the dimension of the dataset (6 rows, 15 columns)
```
